```python
import numpy as np

w1,w2,w3=0.2,0.4,0.5
kanto=np.array([74,66,45])
weights=np.array([w1,w2,w3])
print(np.dot(kanto,weights))
```

```python
climate_data=np.array([[74,66,45],
                       [91,88,64],
                       [87,134,58],
                       [102,43,37],
                       [69,96,70]])
print(np.matmul(climate_data,weights))
print(climate_data@weights)
# Mathematics: np.sum, np.exp, np.round, arithmetic operators
# Array Manipulation: np.reshape, np.stack, np.concatenate, np.split
# Linear Algebra: np.matmul, np.dot, np.transpose, np.eigvals
# Statistics: np.mean, np.median, np.std, np.max
```
